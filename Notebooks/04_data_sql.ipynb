```python
##we create a new column for risk factor
cursor.execute("alter table loan_data add column risk float")
conn.commit()
```

```python
pd.read_sql("select * from loan_data limit 10",conn)
```

#RISK CALCULATION

```python
cursor.execute("""
update loan_data
set risk =
(
    -- default (25%)
    (case when "default" = 'Y' then 5 ELSE 1 end) * 0.25

    -- percent (25%)
  +(case
        when percent > 0.3 then 3
        when percent > 0.6 then 5
        else 1
    end) * 0.25

    -- grade (15%)
  + (case
        when grade = 'A' then 1
        when grade = 'B' then 2
        when grade = 'C' then 3
        when grade = 'D' then 4
        when grade = 'E' then 5
        when grade = 'F' then 6
        when grade = 'G' then 7
    end) * 0.15

    -- employment (10%)
  + (case
        when emp <= 2 then 4
        when emp <= 9 then 2
        else 1
    end) * 0.10

    -- ownership (10%)
  + (case
        when ownership = 'RENT' then 4
        when ownership = 'MORTGAGE' then 2
        else 1
    end) * 0.10

    -- history (10%)
  + (case
        when history < 2 then 5
        when history <= 7 then 3
        else 1
    end) * 0.10

    -- age (5%)
  + (case
        when age < 25 then 2
        when age <= 50 then 1
        when age <= 60 then 3
        else 4
    end) * 0.05

    -- status (5%)
  + (case
        when status = 1 then 3
        else 1
    end) * 0.05
);


""")
conn.commit()
```
