```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
trial=df.head(10)
trial
plt.hist(x=df['age'],bins=20,)
```

```python
fig=plt.figure(figsize=(18,8))

axis1=fig.add_subplot(1,2,1)
ax1=sns.countplot(y=df['intend'])
for container in ax1.containers:
  ax1.bar_label(container)
plt.title('Intend Count')

axis2=fig.add_subplot(1,2,2)
ax2=sns.countplot(y=df['grade'])
for container in ax2.containers:
  ax2.bar_label(container)
plt.title('Grade Count')
```

```python
sns.kdeplot(x=df['income'],y=df['risk'])
```

```python
fig=plt.figure(figsize=(20,8))

axis1=fig.add_subplot(1,2,1)
ax1=sns.histplot(x=df['income'],bins=5,hue=df['ownership'])
ax1.set_xticks([50000,100000,150000,200000,250000])
for container in ax1.containers:
  ax1.bar_label(container)
plt.title('Income v/s Ownership')

axis2=fig.add_subplot(1,2,2)
ax2=sns.histplot(x=df['income'],bins=20,kde=True)
for container in ax2.containers:
  ax2.bar_label(container)
plt.title('Income Distribution')
plt.legend(title=f'max income={round(df['income'].max(),2)} \n min income={round(df['income'].min(),2)} \n avg income={round(df['income'].mean(),2)}', loc='upper right')
```

```python
ax=sns.barplot(x='grade',y='amount',data=df,hue='default')
plt.title('Amount v/s Grade')
for container in ax.containers:
  ax.bar_label(container,rotation=90,padding=3,fmt='%.0f')
```
